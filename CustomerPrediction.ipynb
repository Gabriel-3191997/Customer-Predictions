```python
import pandas as pd

# load dataset
data = pd.read_csv('telco-customer-churn.csv')

# create dataframe 
x = pd.DataFrame(data)

# drop columns 
x = x.drop(columns=[
    'SeniorCitizen', 
    'Partner', 
    'Dependents', 
    'Churn', 
    'PhoneService', 
    'InternetService', 
    'OnlineSecurity', 
    'MultipleLines', 
    'DeviceProtection', 
    'PaperlessBilling',  
    'StreamingMovies', 
    'StreamingTV', 
    'PaymentMethod',
    'OnlineBackup',
    'TechSupport'
], 
           
errors='ignore'

)


# remove Duplicate rows
x.drop_duplicates(inplace=True)


# handle missing values
a = x.dropna()


# display first rows
print(a.head())


# save to file 
a.to_csv("telco-customer-churn.csv", index=False)
```
